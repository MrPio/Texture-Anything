````python
%load_ext autoreload
%autoreload 2
import sys
import pandas as pd
from tqdm import tqdm
from pathlib import Path
from shutil import copy

sys.path.append("../..")
from src import *
````

Expected structure:

```sh
/dataset
  /diffuse
    - diffuse_001.png
    - texture_001.txt
    - ...
  /uv
    - diffuse_001.png
    - ...
```

````python
OUTPUT_DIR = Path("dataset")
(OUTPUT_DIR/"diffuse").mkdir(exist_ok=True, parents=True)
(OUTPUT_DIR/"uv").mkdir(exist_ok=True)
TRAINSET_DIR = Path("../dataset/train")
trainset = pd.read_json("../dataset/train/metadata.jsonl", orient="records", lines=True)
````

````python
for _, sample in tqdm(trainset.iterrows()):
    copy(TRAINSET_DIR / sample.uv_file_name, OUTPUT_DIR / sample.uv_file_name)
    copy(TRAINSET_DIR / sample.diffuse_file_name, OUTPUT_DIR / sample.diffuse_file_name)
    (OUTPUT_DIR / sample.diffuse_file_name).with_suffix(".txt").write_text(sample.caption)
````
